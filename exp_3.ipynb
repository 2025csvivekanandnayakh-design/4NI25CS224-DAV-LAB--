```python
import seaborn as sns
from scipy import stats
```

```python
tips=sns.load_dataset("tips")
print(tips["total_bill"].corr(tips["tip"]))
smokers=tips[tips.smoker=="Yes"]["tip"]
non_sm=tips[tips.smoker=="No"]["tip"]
t,p=stats.ttest_ind(smokers,non_sm)
print(f"t={t:.3f},p={p:.3f}")
print("Reject H0?",p<0.05)
```
