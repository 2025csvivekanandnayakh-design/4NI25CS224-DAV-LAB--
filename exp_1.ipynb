```python
import pandas as pd
import seaborn as sns
tips=sns.load_dataset("tips")
print(tips["total_bill"].mean())
print(tips["total_bill"].median())
print(tips["total_bill"].mode()[0])
print(tips["total_bill"].std())
print(tips["total_bill"].var())
print(tips["total_bill"].describe())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips=sns.load_dataset("tips")
tips["tip_pct"]=tips["tip"]/tips["total_bill"]
sns.histplot(tips["tip_pct"],kde=True)
plt.title("distribution of tip%")
plt.show()
print("skew:",tips["tip_pct"].skew())
print("kurtosis:",tips["tip_pct"].kurt())

```
