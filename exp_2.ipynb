```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("/content/student_data.csv")
print("dataset shape:",df.shape)

```

```python
print(df.head())
```

```python
print(df.info())
```

```python
print(df.describe())
```

```python
plt.figure(figsize=(6,4))
sns.histplot(df["G3"], bins=20, kde=True,color="black")
plt.title("Distribution of Final Grade (G3)")
plt.xlabel("Final Grade")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x="studytime",y="G3",data=df, palette="Set2")
plt.title("Final Grade across Study Time Levels")
plt.xlabel("Study Time(1=low,4=high)")
plt.ylabel("Final Grade")
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.scatterplot(x="G1", y="G3", hue="sex", data=df, palette="coolwarm")
plt.title("First Period Grade vs Final Grade")
plt.xlabel("G1 (Period 1 Grade)")
plt.ylabel("G3 (Final Grade)")
plt.show()
```

```python
sns.pairplot(df[["age","studytime","G1","G2","G3","sex"]],hue="sex",palette="Set2")
plt.suptitle("Pairwise Relationship:",y=1.02)
plt.show()
```

```python
numeric_cols=["age","studytime","failures","absences","G1","G2","G3"]
plt.figure(figsize=(8,6))
sns.heatmap(df[numeric_cols].corr(),annot=True,cmap="coolwarm",fmt=".2f")
plt.title("Correlational Heatmap")
plt.show()
```
